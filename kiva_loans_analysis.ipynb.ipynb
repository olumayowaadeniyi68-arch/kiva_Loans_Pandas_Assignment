```python
import pandas as pd
import numpy as np

df = pd.read_csv("kiva_loans.csv", low_memory=False)

print(df.shape)
df.head()
```

```python
# 1. Which countries account for the largest share of Kiva loans in this dataset?
```

```python
country_loan_counts = (
    df["country"]
    .value_counts()
    .rename_axis("country")
    .reset_index(name="number_of_loans")
)

country_loan_counts["share_of_all_loans_percent"] = (
    country_loan_counts["number_of_loans"] / len(df) * 100
)

country_loan_counts.head(10)
```

The country with the largest share of Kiva loans as seen in this dataset is Philippines, with 160441 number of loans, or about 23.90% of all loans. The next largest countries are Kenya and El Salvador. This means that Kiva lending in this dataset is concentrated in a small number of countries rather than being evenly spread across every country.

# 2. Which farming activities have the highest average loan amounts?

```python
farming_loans = df[
    df["activity"].str.contains("farm", case=False, na=False)
]

farming_loans["activity"].value_counts()
```

```python
farming_summary = (
    farming_loans
    .groupby("activity", as_index=False)
    .agg(
        number_of_loans=("id", "count"),
        average_loan_amount=("loan_amount", "mean")
    )
    .sort_values("average_loan_amount", ascending=False)
)

farming_summary
```

Farm Supplies recorded a higher average requested loan amount of 787.76 than Farming 708.46. This comparison includes 4,352 Farm Supplies loans and 72,955 Farming loans. The analysis only covers activity names containing “farm”, so it does not represent every agricultural activity performed.

# 3. Which sectors have the highest average loan amounts?

```python
sector_summary = (
    df
    .groupby("sector", as_index=False)
    .agg(
        number_of_loans=("id", "count"),
        average_loan_amount=("loan_amount", "mean")
    )
    .sort_values("average_loan_amount", ascending=False)
)

sector_summary
```

Entertainment has the highest average requested loan amount at 1,673.67, followed by Wholesale at 1,569.72 and Clothing at 1,139.24. However, Entertainment and Wholesale have only 830 and 634 loans respectively, compared with 32,742 Clothing loans. These results compare average loan sizes, not the total amount requested by each sector.

# 4. Are food-sector loans typically larger or smaller than loans in other sectors?

```python
df["food_group"] = df["sector"].apply(
    lambda sector: "Food" if sector == "Food" else "Other sectors"
)

food_comparison = (
    df
    .groupby("food_group", as_index=False)
    .agg(
        number_of_loans=("id", "count"),
        median_loan_amount=("loan_amount", "median")
    )
    .sort_values("median_loan_amount", ascending=False)
)

food_comparison
```

Food-sector loans have a median requested amount of 450, compared with 500 for loans in other sectors combined. This means the typical Food-sector loan is smaller by 50 in this dataset. The comparison includes 136,657 Food loans and 534,548 loans from other sectors.

# 5. What percentage of loans are fully funded in this dataset?

```python
fully_funded_loans = df[
    df["funded_amount"] >= df["loan_amount"]
]

total_loans = len(df)
number_fully_funded = len(fully_funded_loans)

percentage_fully_funded = number_fully_funded / total_loans * 100

print(f"Total loans: {total_loans:,}")
print(f"Fully funded loans: {number_fully_funded:,}")
print(f"Percentage fully funded: {percentage_fully_funded:.2f}%")
```

Of all the 671,205 loans in this dataset, 622,877 approximately (92.80%) have funded amounts that meet or exceed their requested amounts. Therefore, most loans were fully funded at the time represented by the dataset. This does not show us whether the remaining loans received more funding later or not.

# 6. Among the five countries with the most loans, which has the highest percentage of fully funded loans?

```python
top_five_countries = df["country"].value_counts().head(5).index

top_country_loans = df[
    df["country"].isin(top_five_countries)
].copy()

top_country_loans["fully_funded"] = (
    top_country_loans["funded_amount"]
    >= top_country_loans["loan_amount"]
)

country_funding_summary = (
    top_country_loans
    .groupby("country", as_index=False)
    .agg(
        number_of_loans=("id", "count"),
        fully_funded_share=("fully_funded", "mean")
    )
)

country_funding_summary["fully_funded_percentage"] = (
    country_funding_summary["fully_funded_share"] * 100
)

country_funding_summary = (
    country_funding_summary[
        ["country", "number_of_loans", "fully_funded_percentage"]
    ]
    .sort_values("fully_funded_percentage", ascending=False)
)

country_funding_summary
```

Among the five countries with the most loans, the Philippines has the highest percentage of fully funded loans at 98.23%, followed by Cambodia at 96.06%. El Salvador has the lowest percentage among these five countries at 82.06%. These results describe funding recorded in this dataset and do not establish which country has the highest full-funding percentage across all countries.
